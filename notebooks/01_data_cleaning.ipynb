# Data Cleaning & Preparation

This notebook handles data extraction, cleaning, and normalization from Flight Centre annual reports (FY2021-FY2024).

## Objectives
1. Extract financial data from raw annual reports
2. Clean and normalize data for consistency
3. Calculate derived metrics
4. Prepare data for exploratory analysis

```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# Load raw financial data
df = pd.read_csv('../data/raw/flt_financials_2021_2024.csv')
print("Raw data shape:", df.shape)
print("\nFirst few rows:")
print(df.head())
```

## Data Cleaning Steps

```python
# Handle missing values
print("Missing values:")
print(df.isnull().sum())

# Fill missing values with appropriate methods
df = df.fillna(0)
print("\nAfter filling missing values:")
print(df.isnull().sum())
```

```python
# Data type conversion
numeric_cols = df.columns[1:]  # All columns except Fiscal Year
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

print("Data types after conversion:")
print(df.dtypes)
```

```python
# Calculate derived metrics
df['revenue_growth_%'] = df['revenue'].pct_change() * 100
df['pat_growth_%'] = df['profit_loss_after_tax'].pct_change() * 100
df['equity_growth_%'] = df['total_equity'].pct_change() * 100
df['operating_margin_%'] = (df['profit_loss_before_tax'] / df['revenue'] * 100).replace([np.inf, -np.inf], 0)
df['net_margin_%'] = (df['profit_loss_after_tax'] / df['revenue'] * 100).replace([np.inf, -np.inf], 0)
df['roe_%'] = (df['profit_loss_after_tax'] / df['total_equity'] * 100).replace([np.inf, -np.inf], 0)
df['debt_to_equity'] = (df['total_liabilities'] / df['total_equity']).replace([np.inf, -np.inf], 0)

print("Derived metrics calculated:")
print(df[['Fiscal Year', 'revenue_growth_%', 'operating_margin_%', 'net_margin_%', 'roe_%', 'debt_to_equity']])
```

```python
# Save cleaned data
df.to_csv('../data/processed/flt_cleaned_data.csv', index=False)
print("Cleaned data saved to: ../data/processed/flt_cleaned_data.csv")
print("\nFinal dataset shape:", df.shape)
print("\nSummary statistics:")
print(df.describe())
```
